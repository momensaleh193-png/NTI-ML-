```python
def to_celsius(fehrenheit) :
    return (fehrenheit - 32) * 5 / 9

print(to_celsius(100))
print(to_celsius(32))
```

```python
def sum(*nums) :
  total =0
  for num in nums :
    total += num
  return total

sum_all = sum(2,8,4,9,11)
print(sum_all)
```

```python
def details (**info) :
  for key , value in info.items() :
    print(key , " : " , value)

x = details(name = "ahmed" , age = 22 , country = "egypt")
```

```python
def student_report (name,*subjects,**grades) :
  print(f"{name }'s subjects are {subjects}")
  for key , value in grades.items() :
    print(key , ":" , value)

student1 = student_report("momen" , "math" , "physics" , math = 88 , physics=92)
```
