```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("store_sales.csv")
```

```python
print("the first 10 rows : ")
df.head(10)
```

```python
print("the last 5 rows :")
df.tail()
```

```python
print("the shape of data is :")
print(df.shape)

print("\ncolumns")
print(df.columns)

print("\nthe data type are")
print(df.dtypes)
```

```python
df[df["Price"].isna() | df["Rating"].isna()]
```

```python
price_mean = df["Price"].mean()
rating_median = df["Rating"].median()
df_clean = df.copy()
df_clean["Price"] = df_clean["Price"].fillna(price_mean)
df_clean["Rating"] = df_clean["Rating"].fillna(rating_median)
```

```python
print(df_clean["Price"].isna().sum())
print(df_clean["Rating"].isna().sum())
```

```python
df_clean["Date"] = pd.to_datetime(df_clean["Date"] , format = "mixed")
```

```python
df_clean[(df_clean["Rating"] > 5) | (df_clean["Rating"] < 1 )]
```

```python
df_clean.loc[(df_clean["Rating"] > 5) | (df_clean["Rating"] < 1 ) , "Rating"] = None
df_clean["Rating"] = df_clean["Rating"].fillna(rating_median)
```

```python
df_clean[df_clean.duplicated()]
df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()
```

```python
print(df_clean.isna().sum())
print("\nthe new shape is : " , df_clean.shape)
```

```python
df_clean.groupby("Category")["Price"].mean()
```

```python
df_clean.groupby("Region")["Quantity"].sum()
```

```python
plt.plot(df_clean["Price"].head(30))
plt.show()
```

```python
plt.plot(df_clean["Price"].head(30))
plt.xlabel("Order")
plt.ylabel("Price")
plt.title("Price Changes")
plt.show()
```

```python
plt.scatter(df_clean["Duration_Min"] , df_clean["Price"])

plt.xlabel("Duration (Minutes)")
plt.ylabel("Price")
plt.title("Duration vs Price")


plt.show()
```

```python
plt.hist(df_clean["Rating"])

plt.xlabel("Rating")
plt.ylabel("Number of Orders")
plt.title("Rating Distribution")

plt.show()
```

```python
category_counts = df_clean["Category"].value_counts()

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%"
)

plt.title("Orders by Category")

plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(8,8))

ax[0, 0].plot(df_clean["Price"].head(30))
ax[0, 0].set_title("Price")

ax[0, 1].hist(df_clean["Rating"])
ax[0, 1].set_title("Rating")

ax[1, 0].scatter(
    df_clean["Duration_Min"],
    df_clean["Price"]
)
ax[1, 0].set_title("Duration and Price")

category_count = df_clean["Category"].value_counts()

ax[1, 1].pie(
    category_count,
    labels=category_count.index,
    autopct="%1.1f%%"
)
ax[1, 1].set_title("Category")

plt.show()
```
